```python
from sklearn import svm
import numpy as np
import matplotlib.pyplot as plt

X = np.array([[0,0],[0.1,0.4], [1,1],[1,1.2]])
y = np.array([0,0,1,1])
clf = svm.SVC()
print(clf.fit(X,y))


#print(clf.predict([[2,2],[1,0],[0,-1]]))



z1 = np.linspace(0, 1, 21)
z2 = np.linspace(0, 1, 21)
Z1, Z2 = np.meshgrid(z1,z2)
print('Z1', Z1.ravel().shape,'Z2',Z2.ravel().shape)
Z3 = np.c_[Z1.ravel(),Z2.ravel()]
print(Z3.shape)
y_mesh = clf.predict(Z3)
y_mesh = y_mesh.reshape(Z1.shape)
print(y_mesh)


plt.contourf(Z1, Z2, y_mesh, cmap="coolwarm")
plt.scatter(X[:,0],X[:,1], c=y, cmap="coolwarm")
plt.xlim(-1,2)
plt.ylim(-1,2)
plt.show()

```

```python
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()

x = data.data
y = data.target

#print(x.shape)
x = x[:,0:2] #Want to try see how this looks graphically for only 2 parameters in the breast cancer data ceel
#print(x.shape)

clf = svm.SVC()
clf.fit(x,y)

x0_min, x0_max = x[:,0].min()-1 , x[:,0].max()+1
x1_min, x1_max = x[:,1].min()-1 , x[:,1].max()+1



z1 = np.linspace(x0_min, x0_max, 21)
z2 = np.linspace(x1_min, x1_max, 21)
Z1, Z2 = np.meshgrid(z1,z2)
Z3 = np.c_[Z1.ravel(),Z2.ravel()]

y_mesh = clf.predict(Z3)
y_mesh = y_mesh.reshape(Z1.shape)



plt.contourf(Z1, Z2, y_mesh, cmap="coolwarm", alpha=0.5)
plt.scatter(x[:,0] , x[:,1], c=y, cmap="coolwarm")
plt.xlim(x0_min, x0_max)
plt.ylim(x1_min, x1_max)
plt.show()

```
